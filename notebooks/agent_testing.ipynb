```python
import pandas as pd
from sklearn.metrics import accuracy_score
```

```python
ground_truth_1 = pd.read_csv('../data/ground_truth.csv')
ground_truth_2 = pd.read_csv('../data/ground_truth_pt2.csv')

gt_df = pd.concat([ground_truth_1, ground_truth_2], ignore_index=True)
```

```python
gt_df.head()
```

```python
agent_test_1 = pd.read_csv('../data/agent_results.csv')
```

```python
gt_df.link.equals(agent_test_1.link)
```

```python
freq_acc  = (gt_df["frequency_heuristic"] == agent_test_1["frequency_heuristic"]).mean()
mal_acc   = (gt_df["malicious_account"] == agent_test_1["malicious_account"]).mean()
sens_acc  = (gt_df["sensationalism"] == agent_test_1["sensationalism"] ).mean()
naive_acc = (gt_df["naive_realism"] == agent_test_1["naive_realism"]).mean()
```

```python
print("Frequency heuristic accuracy:", freq_acc)
print("Malicious account accuracy:", mal_acc)
print("Sensationalism accuracy:", sens_acc)
print("Naive Realism accuracy:", naive_acc)
```
