```python
import pandas as pd
```

## Ground Truth

```python
gt_df = pd.read_csv('../data/ground_truth_full.csv')
```

```python
gt_df.head()
```

## Base Prompt

model = gemini-3-flash-preview

```python
base = pd.read_csv('../data/base_results.csv')
base.head()
```

```python
base_freq_acc  = (gt_df["frequency_heuristic"] == base["frequency_heuristic"]).mean()
base_mal_acc   = (gt_df["malicious_account"] == base["malicious_account"]).mean()
base_sens_acc  = (gt_df["sensationalism"] == base["sensationalism"] ).mean()
base_naive_acc = (gt_df["naive_realism"] == base["naive_realism"]).mean()
```

```python
print("Frequency heuristic accuracy:", base_freq_acc)
print("Malicious account accuracy:", base_mal_acc)
print("Sensationalism accuracy:", base_sens_acc)
print("Naive Realism accuracy:", base_naive_acc)
```
